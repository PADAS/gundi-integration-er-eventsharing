```python
import re
import json

def fill_in_server_defaults(cls, server: str):
        server = re.sub("\/+$", "", server)
        if not(server.startswith("http")):
            server = "https://" + server
        if ("/api/v" not in server):
            server = server + "/api/v1.0"
        return server
```
